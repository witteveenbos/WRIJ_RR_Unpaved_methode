### Selecteer de gebieden en scenarios die gedraaid moeten worden

Ook wordt hier de periode waarover de modellen gerdraaid moeten worden aangegeven.

```python
main_dir = Path("D:\\153961_Oude_IJssel")
dir_model_basis = Path(main_dir, "WRIJ_RR_Unpaved_methode_03_modellen")

# run_names = [
#     "OY_0_kD5_L4",
#     "OY_1_kD20_L4",
#     "OY_2_kD20_L2",
#     "OY_3_kD20_L4_W",
#     "OY_4_kD20_L4_W_InfMax",
#     "OY_5_kD20_L4_W_InfMax_InitGWS",
# ]

# run_name, scenario = "OY_5_kD20_L4_W_InfMax_InitGWS_20140101_20160930", "REF"
run_names = ["OY_5_kD20_L4_W_InfMax_InitGWS_20140101_20160930"]

# # run_name, scenario = "OY_5_kD20_L4_W_InfMax_InitGWS_20130101_20160930", "REF"
# run_names = ["OY_5_kD20_L4_W_InfMax_InitGWS_20130101_20160930"]

# # run_name, scenario = "OY_5_kD20_L4_W_InfMax_InitGWS_20120401_20160930", "REF"
# run_names = ["OY_5_kD20_L4_W_InfMax_InitGWS_20120401_20160930"]
```

```python
simulations_total = pd.DataFrame()

for run_name in run_names:
    for selectie_gebied in selectie_gebieden:
        for scenario in scenarios:
            simulaties = pd.DataFrame()
            
            simulaties["start_date"] = date_range
            simulaties["end_date"] = simulaties["start_date"].shift(-1)
            simulaties.loc[simulaties.index[-1],"end_date"] = pd.to_datetime(end_date)
            simulaties["seizoen"] = (seizoenen * 100)[:len(date_range)]
            simulaties["scenario"] = scenario
            simulaties["gebied"] = selectie_gebied
            simulaties["run_name"] = run_name
            simulaties["restart_in"] = [0] + [1] * len(simulaties.index[1:])
            simulaties["model_name"] = simulaties.apply(lambda x: f"{run_name[:4]}_gebied{selectie_gebied}_{scenario}_rr_{pd.to_datetime(x.start_date).strftime('%Y%m%d')}_{pd.to_datetime(x.end_date).strftime('%Y%m%d')}", axis=1)

            simulations_total = pd.concat([simulations_total, simulaties], axis=0).reset_index(drop=True)

simulations_total
```

### Hier worden de geselecteerde modellen gedraaid

Ook worden restartbestanden doorgegeven, zodat opeenvolgende modelruns op elkaar kunnen aansluiten.

```python
import time

def calculate_simulation_train(simulations_group):
    simulations_group = simulations_group.reset_index(drop=True)
    for index, simulatie in simulations_group.iterrows():
        print(f"START: {str(simulatie.run_name)} - gebied{str(simulatie.gebied)} - {simulatie.scenario} - {simulatie.model_name} [{index}]")
        dir_model = Path(dir_model_basis, simulatie.run_name, f"gebied_{simulatie.gebied}", simulatie.scenario)

        # copy restart-file from previous run
        if index > 0:
            unpaved_restart_out = Path(dir_model, simulations_group.loc[index-1, "model_name"], "rr", "RSRR_OUT")
            unpaved_restart_in = Path(dir_model, simulations_group.loc[index, "model_name"], "rr", "RSRR_IN")
            shutil.copy(unpaved_restart_out, unpaved_restart_in)

        # print(f"RUN: {str(simulatie.run_name)} - gebied{str(simulatie.gebied)} - {simulatie.scenario} - {simulatie.model_name}")
        workdir = Path(dir_model, simulations_group.loc[index, "model_name"])
        batfile = workdir / "run.bat"
        result = subprocess.run(
            ["cmd.exe", "/c", str(batfile.resolve())],
            cwd=str(workdir),
            capture_output=True,
            text=True
        )

        # print(str(simulatie.gebied) + " - " + simulatie.scenario + " - " + simulatie.model_name)
        # print("STDOUT:")
        # print(result.stdout)
        print(f"FINISH: {str(simulatie.gebied)} - {simulatie.scenario} - {simulatie.model_name}")
```

```python
import asyncio
from concurrent.futures import ThreadPoolExecutor

async def main():
    loop = asyncio.get_running_loop()
    with ThreadPoolExecutor() as pool:
        results = await asyncio.gather(
            *(
                loop.run_in_executor(pool, calculate_simulation_train, simulations_group.copy())
                for _, simulations_group in simulations_total.groupby(["run_name", "gebied", "scenario"])
            )
        )
    return results
results = await main()
```
